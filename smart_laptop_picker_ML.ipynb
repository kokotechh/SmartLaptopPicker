```python
# Step 1 - Load the Data: we downloaded the dataset from Kaggle.com
# then we tell Python to read the CSV file and preview the first 5 rows.

import pandas as pd

df = pd.read_csv(r"C:\Users\KOKO\Desktop\laptops.csv")
df.head()
```

```python
# Step 3 - Create Category Labels: we create a new column called 'category'
# by using np.select() to classify each laptop based on its gpu_type, ram_memory and processor_tier.
# categories: High Gaming / Gaming / Gaming-Office / Office

import numpy as np
conditions = [
     (df["gpu_type"] == "dedicated")&(df["ram_memory"]>=32)&(df['processor_tier'].isin(['core i9', 'ryzen 9', 'core i7', 'ryzen 7'])),
     (df["gpu_type"] == "dedicated")&(df["ram_memory"]>=16)&(df["processor_tier"].isin(['core i7', 'core i5', 'ryzen 5', 'ryzen 7'])),
     (df["gpu_type"] == "dedicated")&(df["ram_memory"]>=8)&(df["processor_tier"].isin(['core i5', 'ryzen 5', 'core i3', 'ryzen 3'])),
 ]
 
choices = [
    'High Gaming',
    'Gaming',
    'Gaming-Office']

df["category"] = np.select(conditions, choices, default="Office")

df["category"].value_counts()
```

```python
# Step 4 - Drop Useless Columns: we remove columns that don't help the model predict better.

df = df.drop(columns=["index", "Model", "secondary_storage_type", "secondary_storage_capacity", "is_touch_screen", "brand"])

df.shape
```

```python
# Step 5 - One Hot Encoding: ML models can't read text values,
# so we convert all text columns to numbers using pd.get_dummies().

df = pd.get_dummies(df, columns=['processor_brand', 'processor_tier', 'gpu_brand', 'gpu_type', 'OS', "year_of_warranty", "primary_storage_type" ])

df.shape
```

```python
# Step 6 - Split Data: we split the data into training (80%) and testing (20%).
# X = features (the questions), y = category (the answer).

from sklearn.model_selection import train_test_split
X = df.drop(columns=["category"])
Y = df["category"]

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

print(X_train.shape)
print(X_test.shape)
```

```python
# Step 7 - Train the Model: we use Random Forest Classifier and feed it
# X_train (features) and y_train (answers) so it can learn.

from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
model.fit(X_train, y_train)
```

```python
# Step 8 - Evaluate the Model: we use predict() to let the model guess the category 
# of laptops it has never seen before (X_test).
# then we compare its guesses to the real answers (y_test) using accuracy_score.

y_pred = model.predict(X_test)

from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

```python
# we also use classification_report to see how well it performs for each category separately.

from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
